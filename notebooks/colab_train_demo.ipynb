## Step 7: Optional — patch config for Colab path

This cell rewrites only the `data_root` and `output_dir` fields inside the selected config file, so the rest of your YAML stays unchanged.


```python
import yaml

with open(CONFIG_PATH, 'r') as f:
    cfg = yaml.safe_load(f)

cfg['data_root'] = DATA_ROOT
cfg['output_dir'] = f'{PROJECT_ROOT}/outputs'
cfg['num_workers'] = 2

with open(CONFIG_PATH, 'w') as f:
    yaml.safe_dump(cfg, f, sort_keys=False)

print('Config updated for Colab runtime.')
```
